# 📊 Find Maximum Value (Advanced Jupyter Notebook)

```python
# Optional imports
import numpy as np
import pandas as pd
```

```python
def find_max_value(data):
    if not isinstance(data, list):
        print("Error: Input must be a list")
        return None

    if len(data) == 0:
        print("Warning: Empty list")
        return None

    max_val = data[0]

    for i in range(len(data)):
        if data[i] > max_val:
            max_val = data[i]

    return max_val
```

```python
def find_max_builtin(data):
    if len(data) == 0:
        return None
    return max(data)
```

```python
# Input data
data = [85, 92, 78, 96, 88, 76, 95, 89, 91, 84]
print("Input Data:", data)
```

```python
result = find_max_value(data)
print("\nMaximum Value (Custom Logic):", result)
```

```python
result_builtin = find_max_builtin(data)
print("Maximum Value (Built-in):", result_builtin)
```

```python
np_max = np.max(data)
print("Maximum Value (NumPy):", np_max)
```

```python
df = pd.DataFrame(data, columns=["Values"])
df
```

```python
max_value = find_max_value(data)
print("\nMaximum Value Highlighted:")
print(df[df["Values"] == max_value])
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(data, marker='o')
plt.axhline(y=max_value)
plt.title("Maximum Value Visualization")
plt.xlabel("Index")
plt.ylabel("Values")
plt.show()
```
